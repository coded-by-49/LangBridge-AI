```python
import transformers.pytorch_utils

if not(hasattr(transformers.pytorch_utils,"find_pruneable_heads_and_indices")):
    try:
        from transformers.utils.modeling_utils import find_pruneable_heads_and_indices
        transformers.pytorch_utils.find_pruneable_heads_and_indices = find_pruneable_heads_and_indices
        print("\nSuccessfully loaded from modeling_utils into pytorch_utils")
    except ImportError:
        print("there was an error loading it from modeling_utils")
else:
    print("loaded form transformers.pytorch_utils")
    
```

```python
import json
from comet import download_model, load_from_checkpoint

model_path = download_model("masakhane/africomet-mtl")
model = load_from_checkpoint(model_path)
```

```python
print(hasattr(transformers.pytorch_utils,"is_flax_available"))
```

```python

with open("/kaggle/input/datasets/henryaloh/nllb-600-distilled-eval-dataset/nllb_600_distilled_eval_dataset/nllb_igbo_eng_eval/nllb_igbo_eng_comet_eval_data_0_100.json", "r", encoding="utf-8") as datafile:
    data = json.load(datafile)
    
model_output = model.predict(data, batch_size=8, gpus=1)


print(model_output.system_score) 
```
